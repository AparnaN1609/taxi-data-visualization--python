#### Convert pickup column into datetime format

```python
df['pickup'] = pd.to_datetime(df['pickup'], errors='coerce')
df['pickup']
```

#### Remove rows with missing values in important columns

```python
df.dropna(subset=[
    'pickup',
    'fare',
    'distance',
    'tip',
    'tolls',
    'total',
    'payment',
    'pickup_borough',
    'pickup_zone'
], inplace=True)
```

#### Identify numerical columns & Fill missing values using median

```python
numerical = df.select_dtypes(include=np.number).columns
```

```python
for col in numerical:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())
```

#### Identify categorical columns & Fill missing values using mode

```python
categorical = df.select_dtypes(include=['object', 'category', 'str']).columns
```

```python
for col in categorical:
    if df[col].isnull().sum() > 0:
        mode_value = df[col].mode()

        if not mode_value.empty:
            df[col] = df[col].fillna(mode_value[0])
```

### Visualizations Using Matplotlib / Pandas Plot

#### Line Chart – Fare Over Time

```python
df = df.sort_values(by='pickup')
df
```

```python
plt.figure(figsize=(12,6))
plt.plot(df['pickup'], df['fare'],
         marker='.',
         linestyle='-',
         color='black')

plt.title("Fare Over Time",
          fontsize=15,
          fontweight='bold')

plt.xlabel("Pickup Timestamp")
plt.ylabel("Fare")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```

#### Bar Chart – Total Fare by Pickup Borough

##### Grouping the data and Create the graph

```python
borough_fare = df.groupby('pickup_borough')['fare'].sum()
borough_fare
```

```python
plt.figure(figsize=(10,5))
plt.bar(borough_fare.index,
        borough_fare.values,
        color='skyblue',
        edgecolor='black')
plt.title("Total Fare by Pickup Borough",
          fontsize=15,
          fontweight='bold')
plt.xlabel("Pickup Borough")
plt.ylabel("Total Fare")
plt.xticks(rotation=30)
plt.grid(axis='y')
plt.show()
```

#### Pie Chart – Distribution of Trips by Payment Method

```python
payment_count = df['payment'].value_counts()
payment_count
```

```python
plt.figure(figsize=(7,7))
plt.pie(payment_count,
        labels=payment_count.index,
        autopct='%1.1f%%',
        startangle=90)
plt.title("Distribution of Trips by Payment Method",
          fontsize=15,
          fontweight='bold')
plt.show()
```

#### Histogram – Distribution of Distance

```python
plt.figure(figsize=(10,5))
plt.hist(df['distance'],
         bins=20,
         color='lightgreen',
         edgecolor='blue')
plt.title("Distribution of Taxi Trip Distance",
          fontsize=15,
          fontweight='bold')
plt.xlabel("Distance")
plt.ylabel("Frequency")
plt.grid(axis='y')
plt.show()
```

#### Box Plot – Tip Distribution by Pickup Borough

```python
df.boxplot(column='tip',
           by='pickup_borough',
           figsize=(10,5))

plt.title("Tip Distribution by Pickup Borough",
          fontsize=15,
          fontweight='bold')

plt.suptitle("")

plt.xlabel("Pickup Borough")
plt.ylabel("Tip Amount")

plt.xticks(rotation=30)

plt.grid(axis='y')

plt.show()
```

### Visualizations Using Seaborn

#### Count Plot – Number of Trips by Pickup Borough

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df,
              x='pickup_borough')
plt.title("Number of Trips by Pickup Borough",
          fontsize=15,
          fontweight='bold')
plt.xlabel("Pickup Borough")
plt.ylabel("Number of Trips")
plt.xticks(rotation=30)
plt.show()
```

#### Scatter Plot – Relationship Between Distance and Fare

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df,
                x='distance',
                y='fare',
                hue='pickup_borough')
plt.title("Relationship Between Distance and Fare",
          fontsize=15,
          fontweight='bold')
plt.xlabel("Distance")
plt.ylabel("Fare")
plt.legend(title="Pickup Borough")
plt.show()
```

#### Heatmap – Correlation Between Numerical Variables

```python
numerical = df[['distance', 'fare', 'tip', 'tolls', 'total']]
numerical.head()
```

```python
correlation = numerical.corr()
correlation
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(correlation,
            annot=True,
            cmap='coolwarm',
            linewidths=0.5)
plt.title("Correlation Between Numerical Variables",
          fontsize=14,
          fontweight='bold')
plt.show()
```

#### Pair Plot – Pairwise Relationships Between Variables

```python
top_zones = df['pickup_zone'].value_counts().head(5).index
top_zones
```

```python
pair_data = df[df['pickup_zone'].isin(top_zones)]
```

```python
pair_data = pair_data[['distance',
                       'fare',
                       'tip',
                       'total',
                       'pickup_zone']]
pair_data.head()
```

```python
sns.pairplot(data=pair_data,
             vars=['distance', 'fare', 'tip', 'total'],
             hue='pickup_zone',
             diag_kind='hist')
plt.show()
```

#### Violin Plot – Fare Distribution by Payment Method

```python
plt.figure(figsize=(10,6))

sns.violinplot(data=df,
               x='payment',
               y='fare')

plt.title("Fare Distribution by Payment Method",
          fontsize=15,
          fontweight='bold')

plt.xlabel("Payment Method")
plt.ylabel("Fare")

plt.show()
```
